```python
# import required packages
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score
import pandas as pd

# Define cols to keep for Model A (Lexical URL Features only, for the web app --user dashboard)
cols_A = ["UsingIP", "LongURL", "ShortURL", "Symbol@", "PrefixSuffix-", "SubDomains", 'HTTPS', "AnchorURL", "LinksInScriptTags", "RequestURL", "ServerFormHandler"]
cols_B = cols_A + ["WebsiteTraffic", "AgeofDomain", "DomainRegLen", "GoogleIndex", "DNSRecording"]

# Import raw data
raw_data = pd.read_csv('/home/azimeh/Desktop/Code/PHISHIELD/ml/notebooks/phishing.csv')

# Create the feature set dataframes for both models
dataframe_A = pd.DataFrame(raw_data[cols_A])
dataframe_B = pd.DataFrame(raw_data[cols_B])

# FIX: Keep target as a 1D Series (using single brackets) so scikit-learn doesn't throw a DataConversionWarning
target = raw_data['class']

# Define function to train model A
def train_modelA():
    # Split dataframe into training and testing sets
    modelA_train, modelA_test, modelA_train_target, modelA_test_target = train_test_split(
    dataframe_A, target, test_size=0.2, random_state=42, stratify=target
)
    # Create Random Forest Classifier model instance
    modelA = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    # Train modelA on the datasets
    modelA.fit(modelA_train, modelA_train_target)
    # Generate predictions to use for precision calculations later
    predictions = modelA.predict(modelA_test)
    # Calculate model metrics (F1, recall score, precision and accuracy)
    precision = precision_score(modelA_test_target, predictions, average='weighted')
    accuracy = modelA.score(modelA_test, modelA_test_target)
    f1 = f1_score(modelA_test_target, predictions, average='weighted')
    recall = recall_score(modelA_test_target, predictions, average='weighted')
    return {"model":modelA, "accuracy":accuracy, "precision":precision, "f1_score":f1, "recall":recall}

# Define function to train model B

def train_modelB():
    # Split dataframes into training and testing sets
    modelB_train, modelB_test, modelB_train_target, modelB_test_target = train_test_split(
    dataframe_B, target, test_size=0.2, random_state=42, stratify=target
)
    # Create Random Forest Classifier model instance
    modelB = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
    # Train modelB on the datasets
    modelB.fit(modelB_train, modelB_train_target)
    # Generate predictions to use for precision calculations
    predictions = modelB.predict(modelB_test)
    # Calculate model metrics (F1, recall score, precision and accuracy)
    precision = precision_score(modelB_test_target, predictions, average='weighted')
    accuracy = modelB.score(modelB_test, modelB_test_target)
    f1 = f1_score(modelB_test_target, predictions, average='weighted')
    recall = recall_score(modelB_test_target, predictions, average='weighted')
    return {"model":modelB, "accuracy":accuracy, "precision": precision, "f1_score":f1, "recall":recall}


```
